```python
"""You want to estimate the mean salary of software engineers in a country. 
You take 10 different random samples, each containing 50 engineers, and calculate the sample mean for each.
Plot the distribution of these sample means. How does the Central Limit Theorem explain the shape of this 
sampling distribution, even if the underlying salary distribution is skewed?"""
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
population_size = 100000
population_salaries = np.random.lognormal(mean=10, sigma=0.5, size=population_size)
```

```python
sample_size = 50
num_samples = 10
sample_means = []
for _ in range(num_samples):
 sample = np.random.choice(population_salaries, size=sample_size, replace=False)
 sample_means.append(np.mean(sample))
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
#plt.hist(population_salaries,bins=50, color='skyblue', edgecolor='black')
sns.histplot(population_salaries, kde=True, bins=50, color='skyblue', edgecolor='black')
plt.title("Population Salary Distribution (Skewed)")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.subplot(1,2,2)
#plt.hist(sample_means, bins=5, color='lightgreen', edgecolor='black')
sns.histplot(sample_means, kde=True, bins=8, color='skyblue', edgecolor='black')
plt.title("Distribution of Sample Means (n=50, 10 samples)")
plt.xlabel("Sample Mean Salary")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
print("Population Mean Salary:", np.mean(population_salaries))
print("Sample Means:", sample_means)
print("Mean of Sample Means:", np.mean(sample_means))
```
