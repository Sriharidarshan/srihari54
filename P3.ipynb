```python
"""A dataset contains information about car models, including the engine size (in Liters), fuel efficiency (miles per gallon), and car price. Use a pair plot or correlation matrix to explore the relationships between these variables. Which variables seem to have the strongest relationships, 
and what might be the practical significance of these findings?"""
```

```python
# Importing the required libraries
import pandas as pd
import numpy as np
import seaborn as sns                       #visualisation
import matplotlib.pyplot as plt             #visualisation
```

```python
# Loading the CSV file into a pandas dataframe.
df = pd.read_csv("Automobile_data.csv")
df.head(5)
```

```python
df.columns
```

```python
# Removing irrelevant features
df = df.drop(['index','body-style','engine-type', 'num-of-cylinders'],axis=1)
df.head(5)
```

```python
df.isnull().sum()
```

```python
df['price']=df['price'].fillna(df['price'].mean())
```

```python
# Generate Pair plot
sns.pairplot(df,diag_kind='hist')
plt.show()
```

```python
corr_matrix= df.corr(numeric_only=True)
print("Correlation matrix")
print(corr_matrix)
```

```python
# Plotting a heat map
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix Heatmap')
plt.show()
```
